```python
import uproot as up
import awkward as ak
import matplotlib.pyplot as plt
import vector as vec
import numpy as np
import pandas as pd

path = "/home/usuario/Madgraph/projects/gg_hh-LQs/"
signal = up.open(path + "signal/Events/run_01_decayed_1/tag_1_delphes_events.root")
sm = up.open("/home/usuario/Madgraph/projects/gg_hh-sm/Events/run_02_decayed_1/tag_1_delphes_events.root")

signal_tree = signal["Delphes"]
sm_tree = sm["Delphes"]
```

```python
def print_event_tree(signal_tree, evt):

    pid = signal_tree["Particle.PID"].array()
    m1  = signal_tree["Particle.M1"].array()
    m2  = signal_tree["Particle.M2"].array()
    d1  = signal_tree["Particle.D1"].array()
    d2  = signal_tree["Particle.D2"].array()

    pdg = {
         25: "H",
         21: "g",
          5: "b",
         -5: "b~",
         15: "τ−",
        -15: "τ+",
         22: "γ",
         23: "Z",
         24: "W+",
        -24: "W-",
         11: "e−",
        -11: "e+",
         13: "μ−",
        -13: "μ+",
         12: "νe",
        -12: "ν̄e",
         14: "νμ",
        -14: "ν̄μ",
         16: "ντ",
        -16: "ν̄τ",
    }

    print(f"{'Idx':>4} {'Partícula':>10} {'Madre 1':>15} {'Madre 2':>15} {'Hija 1':>15} {'Hija 2':>15}")

    for i in range(len(pid[evt])):

        # ---------- Nombre de la partícula ----------
        part = pdg.get(int(pid[evt][i]), str(int(pid[evt][i])))

        # ---------- Madre 1 ----------
        if m1[evt][i] == -1:
            mother1 = "-"
        else:
            idx = int(m1[evt][i])
            mother1 = f"{pdg.get(int(pid[evt][idx]), pid[evt][idx])} ({idx})"

        # ---------- Madre 2 ----------
        if m2[evt][i] == -1:
            mother2 = "-"
        else:
            idx = int(m2[evt][i])
            mother2 = f"{pdg.get(int(pid[evt][idx]), pid[evt][idx])} ({idx})"

        # ---------- Hija 1 ----------
        if d1[evt][i] == -1:
            daughter1 = "-"
        else:
            idx = int(d1[evt][i])
            daughter1 = f"{pdg.get(int(pid[evt][idx]), pid[evt][idx])} ({idx})"

        # ---------- Hija 2 ----------
        if d2[evt][i] == -1:
            daughter2 = "-"
        else:
            idx = int(d2[evt][i])
            daughter2 = f"{pdg.get(int(pid[evt][idx]), pid[evt][idx])} ({idx})"

        print(f"{i:4d} {part:>10} {mother1:>15} {mother2:>15} {daughter1:>15} {daughter2:>15}")
```

```python
print_event_tree(signal_tree, 3)
```
